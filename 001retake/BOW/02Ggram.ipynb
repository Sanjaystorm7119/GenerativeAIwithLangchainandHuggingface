An n-gram is a contiguous sequence of n items (usually words or characters) from a given text or speech.

```python
#data cleaning and preprocessing
import nltk , re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import pandas as pd
messages = pd.read_csv("../../spamclassifier/SMSSpamCollection",sep='\t',names=["label","message"])
messages
print(messages.head())
pstemmer = PorterStemmer()

corpus = []
for i in range(0,len(messages)):
    review = re.sub("[^a-zA-Z]"," ",messages['message'][i]).lower().split()
    review=[pstemmer.stem(word) for word in review if not word not in stopwords.words('english')]
    review="".join(review)
    corpus.append(review)
corpus


```

```python
from sklearn.feature_extraction.text import CountVectorizer

# cv = CountVectorizer(max_features=100)

# for binary bow 
cv = CountVectorizer(max_features=500,binary=True,ngram_range=(1,3))

x=cv.fit_transform(corpus).toarray()
x
cv.vocabulary_
```

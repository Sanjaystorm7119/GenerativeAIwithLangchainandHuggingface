```python
#stemming , the process of reducing a word to its stem word to the roots of word known as lemma
# should be a list
from nltk import PorterStemmer , word_tokenize  # stem , tokenise
corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''
print(corpus)

words = word_tokenize(corpus)
print(words)
stemmer = PorterStemmer()
for word in words: 
    print(word,"=========>",stemmer.stem(word))
```

```python
# regex stemmer ====> can use regex in stemming to remove prefix suffix

#stemming , the process of reducing a word to its stem word to the roots of word known as lemma
# should be a list
from nltk import RegexpStemmer , word_tokenize  # stem , tokenise

corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''
# print(corpus)

words = word_tokenize(corpus)
regex = RegexpStemmer("ing$|s$|es$|able$|ed$", min=4)

for word in words: 
    print(word,"=========>",regex.stem(word))
```

```python
#snowball stemmer => better accuracy than porterstemmer , takes a lang as parameter
# should be a list

from nltk import SnowballStemmer , word_tokenize   # stem , tokenise
corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''
print(corpus)

words = word_tokenize(corpus)
print(words)
stemmer = SnowballStemmer("english")
for word in words: 
    print(word,"=========>",stemmer.stem(word))
```

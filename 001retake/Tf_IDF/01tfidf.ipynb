##TF-IDF term frequency (words) , inverse document frequency (sentences)

tf = no. of rep words in a sentence / total number of words in a sentence

idf = logBase_e (no of sentences / no of sentences containing words)

```python
#data cleaning and preprocessing
import nltk , re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd


messages = pd.read_csv("../../spamclassifier/SMSSpamCollection",sep='\t',names=["label","message"])
messages
print(messages.head())
lemmatizer = WordNetLemmatizer()

corpus = []
for i in range(0,len(messages)):
    review = re.sub("[^a-zA-Z]"," ",messages['message'][i]).lower().split()
    review=[lemmatizer.lemmatize(word) for word in review if not word not in stopwords.words('english')]
    review="".join(review)
    corpus.append(review)
corpus


# cv = CountVectorizer(max_features=100)

# for binary bow 
cv = TfidfVectorizer(max_features=500,binary=True,ngram_range=(1,3))

X=cv.fit_transform(corpus).toarray()
# cv.vocabulary_

# import numpy as np
# np.set_printoptions(edgeitems=30,linewidth=3000,formatter=dict(float=lambda x : ".3g" % x))
X
```

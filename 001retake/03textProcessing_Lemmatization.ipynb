```python
#lemmatization : unlike stemmer matching its stem word , lemmatizer matches the lemma, takes a pos tag
#compares with Wordnet corpus using morphy() function
# slow than stemming

from nltk import WordNetLemmatizer , word_tokenize   # stem , tokenise
corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''
print(corpus)

words = word_tokenize(corpus)
# print(words)
lemmatizer = WordNetLemmatizer()
for word in words: 
    print(word,"=========>",lemmatizer.lemmatize(word,pos="v"))


```

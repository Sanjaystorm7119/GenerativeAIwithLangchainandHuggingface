```python
from nltk import word_tokenize, sent_tokenize
corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''
print(corpus)

#para to sentence => sent(ence)_tokenise
sentences = sent_tokenize(corpus)
print(sentences)
print(len(sentences))




```

```python
#printing each line
for sentence in sentences:
    print(sentence)
```

```python
#word tokenise

# para to words , sentences to words
words_from_corpus = word_tokenize(corpus)
print("words_from_corpus")

print(words_from_corpus)

print("words_from_sentences")
for sentence in sentences:
    words_from_sentences = word_tokenize(sentence)
    print(words_from_sentences)
```

```python
#word punct tokenise => punctutations are also split as separate words
from nltk import wordpunct_tokenize
# para to words , sentences to words
words_from_corpus = wordpunct_tokenize(corpus)
print("words_from_corpus")

print(words_from_corpus)

print("words_from_sentences")
for sentence in sentences:
    words_from_sentences = wordpunct_tokenize(sentence)
    print(words_from_sentences)
```

```python
#treebank tokenise => apostrophe like 's are separatesd as one word , full stop (.) of last sentence is considered as separate word

from nltk import TreebankWordTokenizer
# para to words , sentences to words
tokenizer = TreebankWordTokenizer()
words_from_corpus = tokenizer.tokenize(corpus)
print("words_from_corpus")

print(words_from_corpus)

print("words_from_sentences")
for sentence in sentences:
    words_from_sentences = tokenizer.tokenize(sentence)
    print(words_from_sentences)

```

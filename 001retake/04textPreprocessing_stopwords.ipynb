```python

from nltk.corpus import stopwords
from nltk import PorterStemmer , word_tokenize ,sent_tokenize  # stem , tokenise
corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''
# print(corpus)
stemmer=PorterStemmer()
stopwords = set(stopwords.words("english"))
sentences =sent_tokenize(corpus)
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    filtered_words = [word for word in words if word.lower() not in set(stopwords) and word.isalpha()]
    stemmed_word = [stemmer.stem(word) for word in filtered_words]
    sentences[i] = " ".join(stemmed_word)
    # print(filtered_words)
    # print(stemmed_word)
    print(sentences[i])


```

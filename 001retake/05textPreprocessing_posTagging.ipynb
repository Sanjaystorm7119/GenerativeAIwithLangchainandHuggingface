```python
import nltk
from nltk.corpus import stopwords
from nltk import PorterStemmer, word_tokenize, sent_tokenize

# Make sure these are downloaded
# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('averaged_perceptron_tagger')

corpus = '''The sky's shifted from a soft amber to a deep violet as the sun dipped behind the hills, casting long shadows across the quiet valley. A gentle breeze rustled through the tall grass, carrying with it the faint scent of lavender and pine. Somewhere in the distance, a lone bird called out, its song echoing briefly before fading into the hush of dusk. For a moment, everything felt suspended—time slowed, and the world held its breath, as if nature itself was pausing to admire the view.'''

stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

sentences = sent_tokenize(corpus)

# To store stemmed, filtered word lists
stemmed_sentences = []

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    filtered_words = [word for word in words if word.lower() not in stop_words and word.isalpha()]
    stemmed_words = [stemmer.stem(word) for word in filtered_words]
    
    # Store for pos_tag_sents
    stemmed_sentences.append(stemmed_words)
    
    # POS tagging individual sentence
    pos = nltk.pos_tag(stemmed_words)
    print(f"\nPOS tags for sentence {i + 1}:")
    print(pos)

# POS tagging for all sentences at once
print("\nPOS tags for all sentences:")
possent = nltk.pos_tag_sents(stemmed_sentences)
print(possent)
```
